```python
import numpy as np
import pandas as pd

np.random.seed(42)

n = 2000

age = np.random.randint(18, 45, n)
bmi = np.random.uniform(16, 35, n)
hb = np.random.uniform(7, 14, n)
dietScore = np.random.randint(1, 11, n)
protein_g = np.random.uniform(30, 100, n)
calcium_mg = np.random.uniform(400, 1200, n)
iron_mg = np.random.uniform(5, 25, n)

# Add measurement noise (realistic lab variability)
hb += np.random.normal(0, 0.6, n)
protein_g += np.random.normal(0, 4, n)
iron_mg += np.random.normal(0, 1.5, n)
calcium_mg += np.random.normal(0, 50, n)

deficiency_type = []

for i in range(n):
    iron_def = hb[i] < 11 and iron_mg[i] < 12
    protein_def = protein_g[i] < 50
    calcium_def = calcium_mg[i] < 600
    poor_diet = dietScore[i] < 4

    risk_count = sum([iron_def, protein_def, calcium_def, poor_diet])

    if risk_count >= 3:
        deficiency_type.append(4)
    elif iron_def:
        deficiency_type.append(1)
    elif protein_def:
        deficiency_type.append(2)
    elif calcium_def:
        deficiency_type.append(3)
    else:
        deficiency_type.append(0)

df = pd.DataFrame({
    "age": age,
    "bmi": bmi,
    "hb": hb,
    "dietScore": dietScore,
    "protein_g": protein_g,
    "calcium_mg": calcium_mg,
    "iron_mg": iron_mg,
    "deficiency_type": deficiency_type
})

# Add 10% label noise
noise_mask = np.random.rand(n) < 0.10
df.loc[noise_mask, "deficiency_type"] = np.random.randint(0, 5, noise_mask.sum())

df.to_csv("../data/nutrient_data.csv", index=False)

print("✅ Noisy synthetic dataset generated (target ~90% accuracy)")
```
